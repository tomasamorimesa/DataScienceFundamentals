#### 8. Train Using Support Vector Machine (SVM)
a) Create a train and test datasets

```python
X = df_iris.drop(['target', 'flower_name'], axis = 'columns')
X.head()
y = df_iris['target']
y.head()
```

#### 9. Use the method ``train_test_split`` to split train / test sizes. Use 20% for the test size)

HINT: Use ``test_size`` parameter of the ``train_test_split`` method to define the test size (ranges from 0 to 1)

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 10, stratify=y)
y_test.value_counts()
```

#### 10. Print the lenght of both training and test sets

```python
print(len(X_train))
print(len(X_test))
print(len(y_train))
print(len(y_test))
print(y_train.value_counts())
print(y_test.value_counts())
```

#### 11. Train the SVM using the sets created before. Test different kernels.

```python
# Support Vector Classifier
from sklearn.svm import SVC

# C = Regularization
svc_algo = SVC(gamma="auto", kernel="rbf", C=1)

# svc_model.fit will train our model with the train subset (X_train, y_train)
model = svc_algo.fit(X_train, y_train)

# Try other kernels ("rbf" - Radial Based Function, "linear" - Linear and "poly" - Polynomial)
```

#### 12. Evaluate the performance. Create a confusion matrix.
HINT: Use the ``X_test``.

```python
# predictions obtained by using our model (using x_test and keeping predictions in a variable y_predict)
y_predict = model.predict(X_test)

# Import metric libraries
from sklearn.metrics import classification_report, confusion_matrix
      
print(y_predict)
print(iris.target_names)

cm = confusion_matrix(y_test, y_predict, labels=[0, 1, 2])
confusion = pd.DataFrame(cm, index=['is_setosa', 'is_versicolor', 'is_virginica'],
                         columns=['predicted_is_setosa','predicted_is_versicolor', 'predicted_is_virginica'])
confusion
```

#### 13. Print a classfication report.
HINT: Use the ``classification_report`` method.

```python
print(classification_report(y_test, y_predict, zero_division=0.0))
```

#### 14. Print the mean accuracy for all the labels 
HINT: You can use the ``score`` method.

```python
model.score(X, y)
```
