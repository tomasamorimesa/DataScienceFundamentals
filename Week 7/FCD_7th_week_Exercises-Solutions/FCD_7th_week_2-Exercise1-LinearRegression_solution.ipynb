# Linear regression - Regression Problem

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.linear_model import LinearRegression
from sklearn import metrics

%matplotlib inline
```

## Load data from a CSV file

#### 1. Read the CSV file "data_lr.csv" and import it to a DataFrame
Confirm the operation by printing the first 5 rows of the dataframe, where:
- column 1: population (x) - in thousands
- column 2: profit (y) - profit

```python
# read data into a DataFrame
data = pd.read_csv('data_lr.csv')
data.head()
```

#### 2. Separate features (X) from target (y)
Create the input variable (X) and the output variable (y).

```python
# create X and y
X = data[['Population']].values  # features
y = data.Profit                  # target
```

#### 3. Print the number of instances (rows) of our dataset

```python
data.shape[0]
```

#### 4. Fit a linear regression model

```python
# instantiate and fit
linreg = LinearRegression()
model = linreg.fit(X, y)
```

#### 5. Obtain and print the coefficients beta0 (intercept) and beta1 (slope) from the model

```python
# print the coefficients
# Beta0
print(linreg.intercept_)
# Beta1
print(linreg.coef_[0])
```

#### 6. Plot data and regression line (Least Squares Line)

```python
# make predictions
preds = model.predict(X)

# first, plot the observed data
#data.plot(kind='scatter', x='Population', y='Profit')
plt.scatter(data['Population'], data['Profit'])

# then, plot the least squares line
#plt.plot(X, preds, c='red', linewidth=2)
plt.plot(data['Population'], preds, c='red', linewidth=2)
```

#### 7. Calculate the score (i.e. the R-squared score)

```python
model.score(X, y)
```

#### 8. Calculate the MAE (Mean Absolute Error) and the MSE (Mean Squared Error)

```python
y_true = y
y_pred = linreg.predict(X)
print('Mean Absolute Error: ', metrics.mean_absolute_error(y_true, y_pred))
print('Mean Squared Error: ', metrics.mean_squared_error(y_true, y_pred))
```
