### Supervised learning

Supervised machine learning algorithms are used when the output data is known. It requires labeled input and output data to train the model. Predictions can be made by learning from a data set. 

Examples of algorithms: Logistic regression, Decision trees, Neural networks.

### Unsupervised learning

Unsupervised machine learning is used when the target output is unknown and the data is unlabeled. It discovers hidden patterns in data. A typical application is clustering data points in different groups, such as populations, which can help with tasks like market segmentation. 

Examples of algorithms: K-means clustering, Hierarchical clustering, DBSCAN (density-based spatial clustering of applications with noise), Principal component analysis (PCA).
