```python
# using nested loops
vowels = "aeiouAEIOU"
new_list = []
vowel_counter = 0

# for elements in ser:
#     for chars in elements:
#         if(chars in vowels):
#             vowel_counter += 1
#     if(vowel_counter >= 2):
#         new_list.append(elements)
#         vowel_counter = 0
# new_list


# creates a list of vowels
vowels = list("aeiou")

#creates temporary list
aux_list = []

#Loops through all words in ser serie
for word in ser:
    #sets the counter to 0
    c = 0
    #loops through each letter in the word
    for letter in list(word.lower()):
        #checks if the letter exists in the vowels list. if so, increases 1 to the vowel counter
        if letter in vowels:
            c += 1
    #checks if there are 2 or more vowels. if so, appends it to the auxiliary list
    if c > 2:
        aux_list.append(word)
aux_list
```
