### Exercise 1

a) Count the number of vowels in the given list of strings. <br><br>
***Recall vowels are the letters 'a', 'e', 'i', 'o', or 'u'. For this exercise don't include 'y' as a vowel. ***

```python
fruits = ['ApplE', 'banana', 'pear', 'orange', 'grape']

# Your code goes here:
number_of_vowels = 0
#vowels = "aeiouAEIOU"
vowels = ['a','e','i','o','u']

for names in fruits:
    for letters in names:
        if(letters in vowels):
            number_of_vowels += 1
print(number_of_vowels)
```

### Exercise 2

a) Create a new list that contains all the unique values in the following lists. <br><br>
***Expected result: [56, 32, 81, 12, 3, 7, 65, 2, 97] ***

```python
a = [56, 32, 81, 12, 3, 7]
b = [65, 81, 56]
c = [2, 2, 97]

# Your code goes here:
new_list = a + b + c
print(new_list)
no_dups_list = []

for element in new_list:
    if(element not in no_dups_list):
        no_dups_list.append(element)
print(no_dups_list)
```

b) Create a new list that contains all the unique elements of the given nested list. <br><br>
***Expected result: [32, 65, 2, 3, 97, 7, 12, 81, 56]***

```python
lst = [[56, 32, 81, 12, 3, 7], [65, 81, 56], [2, 2, 97]]

# Your code goes here:
no_dups_list_2 = []

for lists in lst:
    for elements in lists:
        if(elements not in no_dups_list_2):
            no_dups_list_2.append(elements)
print(no_dups_list_2)
```

Modify the given list to only keep every **odd** value.<br><br>

***Expected result: [1, 3, 5, 7, 9, 11, 13, 15]***

```python
lst = list(range(0, 16))
print(lst)
```

```python
# Run the code above. Your code goes here:
new_list_3 = []

# Primeira forma
for i in lst:
    if (i % 2 == 0):
        new_list_3.append(i)
lst = new_list_3
print("Even numbers: ", lst)

# Segunda forma
lst = list(range(1, 16, 2))
print("Odd numbers:", lst)
```

```python
# Your code goes here:
soma_primos = 0

i = 1
n = 100
```

```python
soma_primos = 0
n = 10

for i in range(2, n):
    # Assume i is prime
    primo = True
    for j in range(2, int(i**0.5) + 1):
        if i % j == 0:
            primo = False
            break
    if primo:
        soma_primos += i

print(soma_primos)
```

b) Create a dictionary that records how frequently the numbers in 'my_list' fall into the following ranges:<br>
* Greater than or equal to 0 and less than 10 -> use the key [0, 10)<br>
* Greater than or equal to 10 and less than 20 -> use the key [10, 20)<br>
* Greater than or equal to 20 and less than 30 -> use the key [20, 30)<br>
* Greater than or equal to 30 and less than 40 -> use the key [30, 40)<br>
* .. and so on up to 100

Expected output:<br>
{'[0, 10)': 6, 
 '[10,20)': 3, 
 '[20,30)': 6,

 '[40,50)': 5,

```python
my_list = [98, 69, 3, 79, 4, 53, 2, 95, 97, 58, 24, 57, 32, 90,
           15, 29, 92, 40, 22, 99, 23, 33, 55, 38, 12, 49, 9, 42,
           7, 76, 36, 6, 86, 82, 28, 71, 30, 11, 67, 68, 65, 45,
           94, 73, 44, 81, 54, 51, 20, 62]
```

```python
# First run the code in the above cell. Your code goes here:
contador_010 = 0
contador_1020 = 0
contador_2030 = 0
contador_3040 = 0
contador_4050 = 0
contador_5060 = 0
contador_6070 = 0
contador_7080 = 0
contador_8090 = 0
contador_90100 = 0

for i in range(0, len(my_list)):
    if (my_list[i] >= 0 and my_list[i] < 10):
        contador_010 += 1
    elif (my_list[i] >= 10 and my_list[i] < 20):
        contador_1020 += 1
    elif (my_list[i] >= 20 and my_list[i] < 30):
        contador_2030 += 1
    elif (my_list[i] >= 30 and my_list[i] < 40):
        contador_3040 += 1
    elif (my_list[i] >= 40 and my_list[i] < 50):
        contador_4050 += 1
    elif (my_list[i] >= 50 and my_list[i] < 60):
        contador_5060 += 1
    elif (my_list[i] >= 60 and my_list[i] < 70):
        contador_6070 += 1
    elif (my_list[i] >= 70 and my_list[i] < 80):
        contador_7080 += 1
    elif (my_list[i] >= 80 and my_list[i] < 90):
        contador_8090 += 1
    elif (my_list[i] >= 90 and my_list[i] < 100):
        contador_90100 += 1
    
outro_dicionario = {'[0, 10)': contador_010, '[10, 20)': contador_1020, '[20, 30)': contador_2030, 
                    '[30, 40)': contador_3040,'[40, 50)': contador_4050, '[50, 60)': contador_5060,
                    '[60, 70)': contador_6070, '[70, 80)': contador_7080, '[80, 90)': contador_8090, 
                    '[90, 100)': contador_90100}
outro_dicionario
```
